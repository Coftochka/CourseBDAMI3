```python
import pandas as pd
import mplfinance as mpf

df = pd.read_parquet("../../src/data/moex_candles/daily/SBER.parquet")

if not isinstance(df.index, pd.DatetimeIndex):
    for col in ("timestamp", "begin", "time", "date"):
        if col in df.columns:
            df = df.set_index(pd.to_datetime(df[col])).drop(columns=[col])
            break
df.index = pd.to_datetime(df.index)
df.index.name = "Date"

df.columns = [c.capitalize() for c in df.columns]
plot_df = df[["Open", "High", "Low", "Close"]].tail(120).copy()

mpf.plot(
    plot_df,
    type="candle",
    style="charles",
    title="SBER — дневные свечи (последние 120 дней)",
    ylabel="Цена, ₽",
    volume=False,
    figsize=(16, 6),
    show_nontrading=False,
)
```
