```python
from dotenv import load_dotenv; 
load_dotenv()
import os
from moexalgo import session, Market, Ticker
import datetime as dt
from tickers import TICKERS
import pandas as pd
import sys

def GetFormattedDate(timestamp : dt.datetime):
    days = str(timestamp.day)
    if (len(days) < 2):
        days = '0' + days
    month = str(timestamp.month)
    if (len(month) < 2):
        month = '0' + month
    year = str(timestamp.year)
    return f"{year}-{month}-{days}"

def TickerFilename1(ticker, timestamp1:dt.datetime):
    return f"{ticker}_{GetFormattedDate(timestamp1)}.csv"

def TickerFilename2(ticker, timestamp1:dt.datetime, timestamp2:dt.datetime):
    return f"{ticker}_{GetFormattedDate(timestamp1)}_{GetFormattedDate(timestamp2)}.csv"

def TickerFilename2str(ticker, timestamp1:str, timestamp2:str):
    return f"{ticker}_{timestamp1}_{timestamp2}.csv"

def LoadTicker(tickerSymb, dateBegin, dateEnd) -> list[pd.DataFrame]:

    ticker = Ticker(tickerSymb)

    dfs = []

    data = ticker.candles(
        start=dateBegin,
        end="today",
        period=1
    )
    begin = data["begin"].min().split(' ')[0]
    print()


    while (begin <= dateEnd): 
        begin0 = begin
        data = ticker.candles(
            start=begin,
            end="today",
            period=1
        )

        begin = data["begin"].max().split(' ')[0]
        dfs.append(data)

        if not (data is None):
            name = f"./dataset/{tickerSymb}/frac_data/{TickerFilename2str(tickerSymb, begin0, begin)}"
            print("loaded", begin0, begin, name)
            data.to_csv(name, index=False)
            
        else:
            print("empty", begin0, begin)
    return dfs



MOEX_TOKEN = os.getenv("MOEX_TOKEN")
session.TOKEN = MOEX_TOKEN
dataBegin = "2005-01-01"
dataEnd   = "2026-03-20"

def MergeDatasets(ticker_symb, df_list):
    combined_df = pd.concat(df_list, ignore_index=True)
    combined_df = combined_df.drop_duplicates(subset=['begin'], keep='first')
    combined_df['begin'] = pd.to_datetime(combined_df['begin'])
    combined_df['end']   = pd.to_datetime(combined_df['end'])

    combined_df.sort_values('begin', ignore_index=True)
    print(combined_df)
    combined_df.to_csv(f"./dataset/{ticker_symb}/{ticker_symb}_BASIC_FULL.csv", index=False)

ticker = "ALRS"
dfs = LoadTicker(ticker, dataBegin, dataEnd)
MergeDatasets(ticker, dfs)
```
