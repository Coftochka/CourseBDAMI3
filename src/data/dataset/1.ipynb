```python
import pandas as pd 
import os 


def CompressData(namein, nameout):
    dypes = {
        "close" : "float32",
        "high" : "float32",
        "low" : "float32", 
        "value" : "float32",
        "volume" : "int64"
    }
    data = pd.read_csv(namein, header=0, parse_dates=["begin", "end"], dtype=dypes)
    data = data.drop(columns=["end"])
    data.to_parquet(nameout, index=False)


def ProcessDir():
    dirs = os.listdir()
    file_in = ""
    for i in dirs:
        if ("BASIC" in i) and ("csv" in i):
            file_in = i
    if (file_in == ""):
        return
    fileout = file_in[:-3] + "parquet"
    print(os.getcwd(), file_in, fileout)
    CompressData(file_in, fileout)



```
